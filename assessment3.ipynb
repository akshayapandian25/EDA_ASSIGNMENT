```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ASSESSMENT 3
# DATA CLEANING AND UNIVARIATE ANALYSIS

# 1. Import Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Gunnels.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully")

# 2. Display First and Last Six Records
print("First Six Records:")
display(df.head(6))

print("Last Six Records:")
display(df.tail(6))

# 3. Total Number of Missing Values
print("Total Missing Values:", df.isnull().sum().sum())

# 4. Display Records Containing At Least One Missing Value
missing_rows = df[df.isnull().any(axis=1)]

print("Rows Containing Missing Values:")
display(missing_rows.head(10))

# 5. Remove Columns Having More Than 70% Missing Values
missing_percentage = df.isnull().mean() * 100

print("Missing Percentage:")
print(missing_percentage)

columns_above_70 = missing_percentage[
    missing_percentage > 70
].index

print("Columns Removed:", list(columns_above_70))

df = df.drop(columns=columns_above_70)

# Identify numerical and categorical columns
numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Numerical Columns:")
print(numeric_columns)

print("Categorical Columns:")
print(categorical_columns)

# 6. Replace Missing Numerical Values With Mean
df_mean = df.copy()

for column in numeric_columns:
    df_mean[column] = df_mean[column].fillna(
        df_mean[column].mean()
    )

print("Mean Imputation Completed")

display(df_mean.head())

# 7. Replace Missing Numerical Values With Median
df_median = df.copy()

for column in numeric_columns:
    df_median[column] = df_median[column].fillna(
        df_median[column].median()
    )

print("Median Imputation Completed")

display(df_median.head())

df_clean = df_median.copy()

# 8. Replace Missing Categorical Values With Mode
for column in categorical_columns:

    if df_clean[column].isnull().any():

        mode_value = df_clean[column].mode().iloc[0]

        df_clean[column] = df_clean[column].fillna(
            mode_value
        )

print("Categorical Missing Values Replaced")

# 9. Identify Duplicate Rows
duplicate_rows = df_clean[
    df_clean.duplicated()
]

print(
    "Number of Duplicate Rows:",
    len(duplicate_rows)
)

display(duplicate_rows.head(10))

# 10. Remove Duplicate Rows
df_clean = df_clean.drop_duplicates().reset_index(
    drop=True
)

print(
    "Rows After Removing Duplicates:",
    len(df_clean)
)

# 11. Identify Column Data Types
print("Column Data Types:")
print(df_clean.dtypes)

df_clean.info()

# 12. Convert Character Variables Into Categories
object_columns = df_clean.select_dtypes(
    include="object"
).columns

for column in object_columns:
    df_clean[column] = df_clean[column].astype(
        "category"
    )

print("Updated Data Types:")
print(df_clean.dtypes)

# 13. Convert Text Values Into Lowercase
for column in object_columns:

    df_clean[column] = (
        df_clean[column]
        .astype(str)
        .str.lower()
    )

print("Text Converted to Lowercase")

display(df_clean.head())

# Select columns for analysis
num_col = "Slope"
cat_col = "Water"

print(
    "Numerical Variable Selected:",
    num_col
)

print(
    "Categorical Variable Selected:",
    cat_col
)

# 14. Detect Outliers Using Boxplot
plt.figure(figsize=(8, 5))

plt.boxplot(
    df_clean[num_col].dropna()
)

plt.title(
    "Boxplot of " + num_col
)

plt.ylabel(num_col)

plt.show()

# 15. Identify Outliers Using IQR Method
Q1 = df_clean[
    num_col
].quantile(0.25)

Q3 = df_clean[
    num_col
].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR

upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean[num_col] < lower_bound)
    |
    (df_clean[num_col] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print(
    "Number of Outliers:",
    len(outliers)
)

display(outliers.head(10))

# 16. Remove Extreme Outliers
df_no_outliers = df_clean[
    (df_clean[num_col] >= lower_bound)
    &
    (df_clean[num_col] <= upper_bound)
].copy()

print(
    "Original Rows:",
    len(df_clean)
)

print(
    "Rows After Outlier Removal:",
    len(df_no_outliers)
)

# 17. Histogram
plt.figure(figsize=(8, 5))

plt.hist(
    df_clean[num_col].dropna(),
    bins=10,
    edgecolor="black"
)

plt.xlabel(num_col)
plt.ylabel("Frequency")

plt.title(
    "Histogram of " + num_col
)

plt.show()

# 18. Histogram With Density Curve
plt.figure(figsize=(8, 5))

plt.hist(
    df_clean[num_col].dropna(),
    bins=10,
    density=True,
    alpha=0.5,
    edgecolor="black"
)

df_clean[
    num_col
].dropna().plot(
    kind="density"
)

plt.xlabel(num_col)

plt.title(
    "Histogram with Density Curve of " + num_col
)

plt.show()

# 19. Violin Plot
plt.figure(figsize=(7, 5))

plt.violinplot(
    df_clean[num_col].dropna(),
    showmeans=True
)

plt.ylabel(num_col)

plt.title(
    "Violin Plot of " + num_col
)

plt.show()

# 20. Dot Plot For Small Numerical Dataset
small_data = df_clean[
    num_col
].dropna().head(20).values

plt.figure(figsize=(9, 5))

plt.plot(
    small_data,
    np.zeros_like(small_data),
    "o"
)

plt.xlabel(num_col)

plt.yticks([])

plt.title(
    "Dot Plot of " + num_col
)

plt.show()

# 21. Frequency Table For Categorical Variable
frequency_table = df_clean[
    cat_col
].value_counts()

print("Frequency Table:")
print(frequency_table)

# 22. Proportion and Percentage of Each Category
proportion = df_clean[
    cat_col
].value_counts(
    normalize=True
)

percentage = proportion * 100

category_summary = pd.DataFrame({
    "Frequency": frequency_table,
    "Proportion": proportion,
    "Percentage": percentage
})

display(category_summary)

# 23. Bar Chart Showing Frequency
plt.figure(figsize=(9, 5))

frequency_table.plot(
    kind="bar"
)

plt.xlabel(cat_col)
plt.ylabel("Frequency")

plt.title(
    "Frequency Bar Chart of " + cat_col
)

plt.xticks(rotation=45)

plt.show()

# 24. Horizontal Bar Chart
plt.figure(figsize=(9, 5))

frequency_table.plot(
    kind="barh"
)

plt.xlabel("Frequency")
plt.ylabel(cat_col)

plt.title(
    "Horizontal Bar Chart of " + cat_col
)

plt.show()

# 25. Percentage Bar Chart
plt.figure(figsize=(9, 5))

percentage.plot(
    kind="bar"
)

plt.xlabel(cat_col)
plt.ylabel("Percentage")

plt.title(
    "Percentage Bar Chart of " + cat_col
)

plt.xticks(rotation=45)

plt.show()

# 26. Pie Chart
plt.figure(figsize=(7, 7))

frequency_table.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.title(
    "Pie Chart of " + cat_col
)

plt.show()

# 27. Lollipop Chart
freq = frequency_table

plt.figure(figsize=(10, 6))

plt.vlines(
    x=range(len(freq)),
    ymin=0,
    ymax=freq.values
)

plt.scatter(
    range(len(freq)),
    freq.values
)

plt.xticks(
    range(len(freq)),
    freq.index.astype(str),
    rotation=45
)

plt.ylabel("Frequency")
plt.xlabel(cat_col)

plt.title(
    "Lollipop Chart of " + cat_col
)

plt.show()

# 28. Dot Chart
freq = frequency_table.sort_values()

plt.figure(figsize=(9, 6))

plt.scatter(
    freq.values,
    range(len(freq))
)

plt.yticks(
    range(len(freq)),
    freq.index.astype(str)
)

plt.xlabel("Frequency")
plt.ylabel(cat_col)

plt.title(
    "Dot Chart of " + cat_col
)

plt.show()

# 29. Density Plot
plt.figure(figsize=(8, 5))

df_clean[
    num_col
].dropna().plot(
    kind="density"
)

plt.xlabel(num_col)

plt.title(
    "Density Plot of " + num_col
)

plt.show()

# 30. Display Distribution of Slope Across Water Groups
groups = (
    df_clean[cat_col]
    .astype(str)
    .unique()
)

for group in groups:

    group_data = df_clean[
        df_clean[
            cat_col
        ].astype(str) == group
    ]

    plt.figure(figsize=(8, 5))

    plt.hist(
        group_data[num_col].dropna(),
        bins=10,
        edgecolor="black"
    )

    plt.title(
        f"{num_col} Distribution for {cat_col} = {group}"
    )

    plt.xlabel(num_col)
    plt.ylabel("Frequency")

    plt.show()
```
