```python
import pandas as pd
import numpy as np

# ASSESSMENT 2
# DATA IMPORT, TRANSFORM AND BASIC STATISTICAL ANALYSIS

# 1. Import Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Gunnels.csv"
df = pd.read_csv(url)

print("Dataset Imported Successfully")

# 2. Display First 10 Records
display(df.head(10))

# 3. Display Last 10 Records
display(df.tail(10))

# 4. Total Number of Rows and Columns
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# 5. Display Names of All Variables
print("Column Names:")
print(df.columns.tolist())

# 6. Display Data Type of Each Variable
print("Data Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

# 7. Display Summary Statistics
display(df.describe(include="all"))

# 8. Number of Missing Values in Each Column
print("Missing Values in Each Column:")
print(df.isnull().sum())

# 9. Total Missing Values in Dataset
total_missing = df.isnull().sum().sum()

print("Total Missing Values:")
print(total_missing)

# 10. Missing Values Column-wise
missing_columnwise = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

display(missing_columnwise)

# 11. Rows Having At Least One Missing Value
missing_rows = df[df.isnull().any(axis=1)]

print("Rows Containing Missing Values:")
display(missing_rows.head(10))

# 12. Remove Rows Having Missing Values
cleaned_dataset = df.dropna()

print("Dataset After Removing Missing Values:")
display(cleaned_dataset.head(10))

df_processed = df.copy()

numeric_columns = df_processed.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = df_processed.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Numerical Columns:")
print(numeric_columns)

print("\nCategorical Columns:")
print(categorical_columns)

# 13. Replace Missing Numerical Values Using Mean
if len(numeric_columns) >= 1:

    mean_column = numeric_columns[0]

    mean_value = df_processed[mean_column].mean()

    df_processed[mean_column] = (
        df_processed[mean_column]
        .fillna(mean_value)
    )

    print("Mean Imputation Applied On:", mean_column)

# 14. Replace Missing Values Using Median
if len(numeric_columns) >= 2:

    median_column = numeric_columns[1]

    median_value = df_processed[median_column].median()

    df_processed[median_column] = (
        df_processed[median_column]
        .fillna(median_value)
    )

    print("Median Imputation Applied On:", median_column)

# 15. Replace Categorical Missing Values Using Mode
if len(categorical_columns) >= 1:

    categorical_column = categorical_columns[0]

    mode_value = df_processed[
        categorical_column
    ].mode()[0]

    df_processed[categorical_column] = (
        df_processed[categorical_column]
        .fillna(mode_value)
    )

    print(
        "Mode Imputation Applied On:",
        categorical_column
    )

# 16. Forward Filling
df_forward = df.copy()
df_forward = df_forward.ffill()

print("Forward Filled Dataset:")
display(df_forward.head(10))

# 17. Backward Filling
df_backward = df.copy()
df_backward = df_backward.bfill()

print("Backward Filled Dataset:")
display(df_backward.head(10))

# 18. Replace Remaining Missing Values With Constant
for column in df_processed.columns:

    if pd.api.types.is_numeric_dtype(
        df_processed[column]
    ):

        df_processed[column] = (
            df_processed[column]
            .fillna(0)
        )

    else:

        df_processed[column] = (
            df_processed[column]
            .fillna("Unknown")
        )

print("Remaining Missing Values:")
print(df_processed.isnull().sum())

# 19. Remove Duplicate Records
print("Number of Duplicate Rows:")
print(df_processed.duplicated().sum())

df_processed = df_processed.drop_duplicates()

print("Duplicate Records Removed")

# 20. Convert Appropriate Variables into Categories
for column in categorical_columns:

    if column in df_processed.columns:

        df_processed[column] = (
            df_processed[column]
            .astype("category")
        )

print(df_processed.dtypes)

# 21. Rename One Column
old_column = df_processed.columns[0]

new_column = old_column + "_New"

df_processed.rename(
    columns={
        old_column: new_column
    },
    inplace=True
)

print("Renamed:")
print(old_column, "->", new_column)

print(df_processed.columns.tolist())

# 22. Create New Derived Column
numeric_columns_new = (
    df_processed
    .select_dtypes(include=np.number)
    .columns
)

if len(numeric_columns_new) >= 2:

    df_processed["Derived_Column"] = (
        df_processed[numeric_columns_new[0]]
        +
        df_processed[numeric_columns_new[1]]
    )

print("Dataset with Derived Column:")
display(df_processed.head(10))

# 23. Summary Statistics
display(
    df_processed.describe(
        include="all"
    )
)

numeric_columns_new = (
    df_processed
    .select_dtypes(include=np.number)
    .columns
)

selected_column = numeric_columns_new[0]

print(
    "Selected Numerical Column:",
    selected_column
)

# 24. Minimum and Maximum Values
minimum = df_processed[
    selected_column
].min()

maximum = df_processed[
    selected_column
].max()

print("Minimum:", minimum)
print("Maximum:", maximum)

# 25. Range
data_range = maximum - minimum

print("Range:", data_range)

# 26. Variance
variance = df_processed[
    selected_column
].var()

print("Variance:", variance)

# 27. Standard Deviation
standard_deviation = df_processed[
    selected_column
].std()

print(
    "Standard Deviation:",
    standard_deviation
)

# 28. Interquartile Range
Q1 = df_processed[
    selected_column
].quantile(0.25)

Q3 = df_processed[
    selected_column
].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```
