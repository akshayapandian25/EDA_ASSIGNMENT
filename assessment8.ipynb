```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from scipy.spatial.distance import pdist

# ASSESSMENT 8
# K-MEANS CLUSTERING

# 1. Load Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Gunnels.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully")

# 2. View Dataset
display(df.head(10))

# 3. Check Missing Values
print("Missing Values:")
print(df.isnull().sum())

# 4. Dependent Variable Distribution
print("Gunnel Distribution:")
print(df["Gunnel"].value_counts())

# 5. Select Only Numerical Features
numeric_data = df.select_dtypes(
    include=np.number
).copy()

print("Numerical Features:")
print(numeric_data.columns.tolist())

# Remove missing values
numeric_data = numeric_data.dropna()

display(numeric_data.head())

# 6. Standardize Data
scaler = StandardScaler()

data_scaled = scaler.fit_transform(
    numeric_data
)

print("Standardized Data:")
print(data_scaled[:5])

# 7. Apply K-Means With K = 2
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(
    data_scaled
)

# 8. Display K-Means Results
kmeans_result = numeric_data.copy()

kmeans_result["Cluster"] = clusters

display(kmeans_result.head(10))

print("Cluster Distribution:")
print(
    kmeans_result[
        "Cluster"
    ].value_counts()
)

# 9. Display Cluster Centers
print("Cluster Centers in Standardized Form:")
print(kmeans.cluster_centers_)

centers_original = scaler.inverse_transform(
    kmeans.cluster_centers_
)

centers_df = pd.DataFrame(
    centers_original,
    columns=numeric_data.columns
)

print("Cluster Centers in Original Scale:")
display(centers_df)

# 10. Calculate WCSS for K = 1 to 10
wcss = []

for k in range(1, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    model.fit(data_scaled)

    wcss.append(
        model.inertia_
    )

print("WCSS Values:")
print(wcss)

# 11. Elbow Plot
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, 11),
    wcss,
    marker="o"
)

plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")

plt.xticks(
    range(1, 11)
)

plt.show()

# 12. Apply K-Means Based on Elbow Recommendation
# Assuming elbow suggests K = 3

final_kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

final_clusters = final_kmeans.fit_predict(
    data_scaled
)

final_result = numeric_data.copy()

final_result["Cluster"] = final_clusters

print("Final K-Means Result:")
display(final_result.head(10))

print("Final Cluster Distribution:")
print(
    final_result[
        "Cluster"
    ].value_counts()
)


# HIERARCHICAL CLUSTERING

# 1. Load Dataset
df_h = pd.read_csv(url)

# 2. View Dataset
display(df_h.head(10))

# 3. Check Missing Values
print("Missing Values:")
print(df_h.isnull().sum())

# 4. Select Only Numerical Attributes
numeric_h = df_h.select_dtypes(
    include=np.number
).dropna()

print("Numerical Attributes:")
print(numeric_h.columns.tolist())

# 5. Standardize Data
scaler_h = StandardScaler()

scaled_h = scaler_h.fit_transform(
    numeric_h
)

print("Standardized Data:")
print(scaled_h[:5])

# 6. Calculate Euclidean Distance
euclidean_distance = pdist(
    scaled_h,
    metric="euclidean"
)

print("First 10 Euclidean Distances:")
print(euclidean_distance[:10])

# 7. Hierarchical Clustering Using Ward Method
hierarchical_model = linkage(
    scaled_h,
    method="ward"
)

print("Hierarchical Clustering Completed")

# 8. Plot Dendrogram
plt.figure(figsize=(12, 6))

dendrogram(
    hierarchical_model
)

plt.title(
    "Hierarchical Clustering Dendrogram"
)

plt.xlabel("Observations")
plt.ylabel("Euclidean Distance")

plt.show()

# 9. Divide Data Into 3 Clusters
hierarchical_clusters = fcluster(
    hierarchical_model,
    t=3,
    criterion="maxclust"
)

hierarchical_result = numeric_h.copy()

hierarchical_result[
    "Cluster"
] = hierarchical_clusters

print("Cluster Assignments:")
display(
    hierarchical_result.head(10)
)

# 10. Number of Samples in Each Cluster
print("Number of Samples in Each Cluster:")

print(
    hierarchical_result[
        "Cluster"
    ].value_counts().sort_index()
)
```
