```python
import pandas as pd
import numpy as np

# ASSESSMENT 1
# MERGING AND DATA TRANSFORMATION


# Create dataframe1 with minimum 10 records
dataframe1 = pd.DataFrame({
    "ProductID": [101,102,103,104,105,106,107,108,109,110],
    "NameofProd": [
        "Laptop", "Mouse", "Keyboard", "Monitor", "Printer",
        "Speaker", "Webcam", "Router", "HardDisk", "Pendrive"
    ],
    "Price": [
        55000, 800, 1500, 12000, 9000,
        2500, 1800, 3000, 4500, 700
    ]
})


# Create dataframe2 with minimum 10 records
dataframe2 = pd.DataFrame({
    "ProductID": [101,102,103,104,105,106,107,108,109,110],
    "Company_Location": [
        "Chennai", "Vellore", "Bangalore", "Mumbai", "Delhi",
        "Hyderabad", "Pune", "Kolkata", "Coimbatore", "Madurai"
    ]
})


# Create dataframe3 with minimum 10 records
dataframe3 = pd.DataFrame({
    "ProductID": [111,112,113,114,115,116,117,118,119,120],
    "NameofProd": [
        "Tablet", "Smartphone", "Charger", "PowerBank", "Headphone",
        "Microphone", "SSD", "Smartwatch", "Camera", "Projector"
    ],
    "Price": [
        25000, 35000, 1200, 2000, 2200,
        3500, 6000, 5000, 45000, 30000
    ]
})


# Display the dataframes
print("DATAFRAME 1")
display(dataframe1)

print("DATAFRAME 2")
display(dataframe2)

print("DATAFRAME 3")
display(dataframe3)

# 1. Append dataframe1 and dataframe3


appended_dataframe = pd.concat(
    [dataframe1, dataframe3],
    ignore_index=True
)

print("APPENDED DATAFRAME")
display(appended_dataframe)

# 2. Concatenate Dataframes


# Vertical concatenation
vertical_concat = pd.concat(
    [dataframe1, dataframe3],
    axis=0,
    ignore_index=True
)

print("VERTICAL CONCATENATION")
display(vertical_concat)


# Horizontal concatenation
horizontal_concat = pd.concat(
    [
        dataframe1.reset_index(drop=True),
        dataframe2["Company_Location"].reset_index(drop=True)
    ],
    axis=1
)

print("HORIZONTAL CONCATENATION")
display(horizontal_concat)


# 3. Merge dataframe1 and dataframe2 using ProductID


merged_dataframe = pd.merge(
    dataframe1,
    dataframe2,
    on="ProductID"
)

print("MERGED DATAFRAME")
display(merged_dataframe)


# 4A. Inner Join


inner_join = pd.merge(
    dataframe1,
    dataframe2,
    on="ProductID",
    how="inner"
)

print("INNER JOIN")
display(inner_join)

# 4B. Left Join


left_join = pd.merge(
    dataframe1,
    dataframe2,
    on="ProductID",
    how="left"
)

print("LEFT JOIN")
display(left_join)

# 4C. Right Join

right_join = pd.merge(
    dataframe1,
    dataframe2,
    on="ProductID",
    how="right"
)

print("RIGHT JOIN")
display(right_join)

# 4D. Full Outer Join

full_outer_join = pd.merge(
    dataframe1,
    dataframe2,
    on="ProductID",
    how="outer"
)

print("FULL OUTER JOIN")
display(full_outer_join)

# PART B - DATA TRANSFORMATION

Akshaya = np.array([
    30, 45, 78, 25, 90,
    56, 67, 34, 82, 100
], dtype=float)

print("Original Vector:")
print(Akshaya)

# 1A. Normalize to [-3, +3]

min_value = np.min(Akshaya)
max_value = np.max(Akshaya)

norm_minus3_plus3 = (
    -3 +
    ((Akshaya - min_value) * 6) /
    (max_value - min_value)
)

print("\nNormalization [-3, +3]")
print(norm_minus3_plus3)


# 1B. Normalize to [-1, +1]


norm_minus1_plus1 = (
    -1 +
    ((Akshaya - min_value) * 2) /
    (max_value - min_value)
)

print("\nNormalization [-1, +1]")
print(norm_minus1_plus1)

# 1C. Min-Max Normalization [0,1]


minmax = (
    (Akshaya - min_value) /
    (max_value - min_value)
)

print("\nMin-Max Normalization [0,1]")
print(minmax)

# 2. Z-Score Transformation


mean_value = np.mean(Akshaya)

std_value = np.std(
    Akshaya,
    ddof=1
)

z_score = (
    Akshaya - mean_value
) / std_value

print("\nMean:")
print(mean_value)

print("\nStandard Deviation:")
print(std_value)

print("\nZ-Score Transformation:")
print(z_score)



# 3. Decimal Scaling Normalization


max_abs_value = np.max(
    np.abs(Akshaya)
)

j = int(
    np.ceil(
        np.log10(max_abs_value + 1)
    )
)

decimal_scaled = Akshaya / (10 ** j)

print("\nDecimal Scaling Value j:")
print(j)

print("\nDecimal Scaling Normalization:")
print(decimal_scaled)
```
